```python
def age_diff(team1,team2):
    idx1 = new_df[new_df['team'] == team1].index[0]
    idx2 = new_df[new_df['team'] == team2].index[0]
    
    
    diff = new_df.loc[idx1, 'avg_squad_age'] - new_df.loc[idx2, 'avg_squad_age']
    return diff
    
```

```python
age_diff('France', 'Senegal')
```

```python
def defense_stability_diff(team1, team2):
    idx1 = new_df[new_df['team'] == team1].index[0]
    idx2 = new_df[new_df['team'] == team2].index[0]
    
    # Calculate average goals conceded per match over the last 5 games
    avg_ga_t1 = new_df.loc[idx1, 'GA'] / 5
    avg_ga_t2 = new_df.loc[idx2, 'GA'] / 5
    
    # A negative difference means Team 1 concedes FEWER goals than Team 2 (better defense)
    return avg_ga_t1 - avg_ga_t2
```

```python
X_list = []
y_list = []

for match in training_fixtures:
    t1 = match['t1']
    t2 = match['t2']
    
    
    match_features = {
        'rating_diff': Rating_Difference(t1, t2),
        'value_diff': value_difference(t1, t2),
        'form_diff': form_diff(t1, t2),
        'age_diff': age_diff(t1, t2),
        'def_stability_diff': defense_stability_diff(t1, t2) 
    }
    X_list.append(match_features)
    
    match_targets = {
        'match_label': match.get('label'), 
        'match_gd': match.get('gd'),       
        'match_gf': match.get('gf'),       
        'match_ga': match.get('ga')        
    }
    y_list.append(match_targets)

X_train = pd.DataFrame(X_list)
y_train = pd.DataFrame(y_list)


X_train
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV


rf_base = RandomForestClassifier(random_state=42)


param_grid = {
    'n_estimators': [10, 30, 50],       # Number of trees in the forest
    'max_depth': [3, 4, 5, None],       # Limit tree depth to prevent memorization
    'min_samples_split': [2, 5],        # Minimum samples required to split a node
    'max_features': ['sqrt', 'log2']    # Number of features considered at each split
}

# 3. Using up Grid Search 
# (cv=3 means it will split your 30 rows into three 10-row chunks to validate)
grid_search = GridSearchCV(
    estimator=rf_base, 
    param_grid=param_grid, 
    cv=3, 
    scoring='accuracy', 
    n_jobs=-1
)


y_label = y_train['match_label']
grid_search.fit(X_train, y_label)


print("Best Parameters Found:")
print(grid_search.best_params_)
print(f"\n Best Cross-Validation Accuracy: {grid_search.best_score_*100:.1f}%")


best_rf_model = grid_search.best_estimator_
```

```python
def predict_match_percentages(team1, team2):
   
    match_features = pd.DataFrame([{
        'rating_diff': Rating_Difference(team1, team2),
        'value_diff': value_difference(team1, team2),
        'form_diff': form_diff(team1, team2),
        'age_diff': age_diff(team1, team2),
        'def_stability_diff': defense_stability_diff(team1, team2)
    }])
    
   
    probabilities = grid_search.predict_proba(match_features)[0]
    
    
    t2_win_pct = probabilities[0] * 100
    draw_pct = probabilities[1] * 100
    t1_win_pct = probabilities[2] * 100
    
    print(f" 2026 World Cup Match Simulation: {team1} vs {team2}")
    print("=" * 50)
    print(f" {team1} Win Probability: {t1_win_pct:.1f}%")
    print(f" Draw Probability: {draw_pct:.1f}%")
    print(f" {team2} Win Probability: {t2_win_pct:.1f}%")
```

```python
def predict_full_match_analytics(team1, team2):

    match_features = pd.DataFrame([{
        'rating_diff': Rating_Difference(team1, team2),
        'value_diff': value_difference(team1, team2),
        'form_diff': form_diff(team1, team2),
        'age_diff': age_diff(team1, team2),
        'def_stability_diff': defense_stability_diff(team1, team2)
    }])
    
    
    base_gd = best_rf_regressor.predict(match_features)[0]
    
    # 3. Simulating the match 1,000 times for the percentages
    t1_wins = 0
    draws = 0
    t2_wins = 0
    simulations = 1000
    
    for _ in range(simulations):
        simulated_gd = np.random.normal(loc=base_gd, scale=1.22) #This takes that exact base_gd anchor point and shakes it up with match-day randomness using your 1.22 MAE score. This creates the realistic win/loss/draw percentages.
        if simulated_gd > 0.25:
            t1_wins += 1
        elif simulated_gd < -0.25:
            t2_wins += 1
        else:
            draws += 1
            
    t1_pct = (t1_wins / simulations) * 100
    draw_pct = (draws / simulations) * 100
    t2_pct = (t2_wins / simulations) * 100
    
    # Generating a realistic exact scoreline using a baseline average of 2.5 goals per game
    # I split the baseline goals using the predicted goal difference
    avg_total_goals = 2.5
    t1_expected_goals = max(0.2, (avg_total_goals + base_gd) / 2)
    t2_expected_goals = max(0.2, (avg_total_goals - base_gd) / 2)
    
    # Sample actual integer goals using the Poisson distribution
    t1_score = np.random.poisson(lam=t1_expected_goals)
    t2_score = np.random.poisson(lam=t2_expected_goals)
    
    
    print("Matchup:", team1, "vs", team2)
    print("---------------------------------")
    print(team1, "Win Probability:", round(t1_pct, 1), "%")
    print("Draw Probability:", round(draw_pct, 1), "%")
    print(team2, "Win Probability:", round(t2_pct, 1), "%")
    print("---------------------------------")
    print("Simulated Scoreline:", team1, t1_score, "-", t2_score, team2)
```
